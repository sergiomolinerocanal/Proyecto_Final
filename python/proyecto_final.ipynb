# Proyecto final Data Analitics

Nos piden que analicemos las ventas de una empresa de franquicias de hostelería.  
Nos entregan una serie de ficheros clasificados en dos carpetas una de "maestros" y otra de "ventas". Vamos a ver lo que contiene cada fichero

```python
import pandas as pd
import numpy as np
import os
```

Abrimos ahora los ficheros para examinar su contenido. Lo haremos en orden alfabético

```python
base = os.path.dirname(os.path.abspath("PROYECTO FINAL")) #Esto hace que nos construya la ruta correctamente en la mayoría de sistemas operativos
ruta_art = os.path.join(base, "data", "maestros", "artículos.xlsx")

df_arti = pd.read_excel(ruta_art, index_col = "id_producto")
df_arti.sample(5)
```

```python
print(f"Nuestro dataframe original de ARTÍCULOS consta de {df_arti.shape[0]} filas y {df_arti.shape[1]} columnas")
print(f"El volumen de datos que contiene nuestro dataframe es: {df_arti.size}")
print(f"Del tota de filas, {df_arti.index.nunique()} son valores únicos")
```

Veamos el tipo de datos que tenemos

```python
df_arti.info()
```

5 columnas de texto descriptivo de cada producto y 2 númericas con el PVP y el coste. Hay dos columnas (Marca y Formato) que no están completas.

```python
ruta_cp = os.path.join(base, "data", "maestros", "codigos_postales.xlsx")

df_cp = pd.read_excel(ruta_cp,index_col = "Código postal")

df_cp.sample(5)
```

```python
print(f"Nuestro dataframe original de CÓDIGOS POSTALES consta de {df_cp.shape[0]} filas y {df_cp.shape[1]} columnas")
print(f"El volumen de datos que contiene nuestro dataframe es: {df_cp.size}")
print(f"Del total de filas, {df_cp.index.nunique()} son valores únicos")
```

El código postal, aunque es un número, en este caso, es mejor que lo tratemos como un texto, ya que no realizaremos ninguna operación aritmética con este dato

```python
df_cp.index = df_cp.index.astype(str)
```

```python
df_cp.info()
```

Vemos que este fichero contiene un total de 76 registros que corresponden a distintos códigos postales con los datos de ubicación geográfica de cada uno y un campo numérico que indica el número de habitantes, que por el volumen de las cifras, debe corresponder a las localidades

```python
ruta_fq = os.path.join(base, "data", "maestros", "franquiciados.xlsx")

df_fq = pd.read_excel(ruta_fq,index_col = "id_franquiciado")
df_fq.sample(5)
```

En este fichero de FRANQUICIADOS tenemos un listado de personas y la descripción de su relación con la franquicia. Tambien vemos que hay 3 columnas vacías que eliminaremos a continuación.

```python
df_fq = df_fq.drop(df_fq.columns[[6, 7, 8, 9]], axis=1)
df_fq.info()
```

```python
print(f"Nuestro dataframe original de FRANQUICIADOS consta de {df_fq.shape[0]} filas y {df_fq.shape[1]} columnas")
print(f"El volumen de datos que contiene nuestro dataframe es: {df_fq.size}")
print(f"Del total de filas, {df_fq.index.nunique()} son valores únicos")
```

Por último, veamos qué contiene el fichero locales

```python
ruta_loc = os.path.join(base, "data", "maestros", "locales.xlsx")

df_loc = pd.read_excel(ruta_loc, index_col = "ID_tienda")
df_loc.sample(5)
```

Este parece ser el fichero principal, contiene la relación de tiendas, su ubicación (que podremos relacionar con el fichero de CP), el ID_franquiciado (que relacionaremos con el fichero FRANQUICIADOS) y algunos otros datos descriptivos de cada local

```python
df_loc.info()
```

```python
df_loc["Cód Postal"] = df_loc["Cód Postal"].astype(str)
```

```python
print(f"Nuestro dataframe original de LOCALES consta de {df_loc.shape[0]} filas y {df_loc.shape[1]} columnas")
print(f"El volumen de datos que contiene nuestro dataframe es: {df_loc.size}")
print(f"Del total de filas, {df_loc.index.nunique()} son valores únicos")
```

Visto que tenemos 3 ficheros que se complentan, vamos a unirlos en un solo DATAFRAME.

```python
df_locales = df_loc.merge(
    df_fq,
    left_on="ID_franquiciado",
    right_index=True,
    how="left"
)
```

```python
df_final = df_locales.merge(
    df_cp,
    left_on="Cód Postal",
    right_index=True,
    how="left"
)
```

```python
df_final.info()
```

```python
print(f"Nuestro dataframe FINAL consta de {df_final.shape[0]} filas y {df_final.shape[1]} columnas")
print(f"El volumen de datos que contiene nuestro dataframe es: {df_final.size}")
print(f"Del total de filas, {df_final.index.nunique()} son valores únicos")
```

Ahora ya tenemos un solo DataFrame con los datos de todas las tiendas, tanto de su descripción geográfica como de la información de cada franquiciado.

Vamos a ver ahora la carpeta de ventas

```python
ruta_2024_S1 = os.path.join(base, "data", "ventas", "ventas_2024_S1.csv")

df_2024S1 = pd.read_csv(ruta_2024_S1)
df_2024S1.sample(5)
```

```python
ruta_2024_S2 = os.path.join(base, "data", "ventas", "ventas_2024_S2.csv")

df_2024S2 = pd.read_csv(ruta_2024_S2)
df_2024S2.sample(5)
```

```python
ruta_2025_S1 = os.path.join(base, "data", "ventas", "ventas_2025_S1.csv")

df_2025S1 = pd.read_csv(ruta_2025_S1)
df_2025S1.sample(5)
```

```python
ruta_2025_S2 = os.path.join(base, "data", "ventas", "ventas_2025_S2.csv")

df_2025S2 = pd.read_csv(ruta_2025_S2)
df_2025S2.sample(5)
```

Los 4 ficheros tienen la misma estructura, cada uno contiene las ventas por artículo, día y local de un semestre y año. Así que a continuación, los uniremos en un solo DataFrame que acumule los dos años de ventas.

```python
ventas = [df_2024S1, df_2024S2, df_2025S1, df_2025S2]
df_ventas = pd.concat(ventas)
df_ventas.info()
```

```python
print(f"Nuestro dataframe de VENTAS consta de {df_ventas.shape[0]} filas y {df_ventas.shape[1]} columnas")
print(f"El volumen de datos que contiene nuestro dataframe es: {df_ventas.size}")
```

```python
#Convertimos el campo "fecha" para poder operar con él correctamente
df_ventas["fecha"] = pd.to_datetime(df_ventas["fecha"], format="%Y-%m-%d")
```

Por lo que vimos anteriormente, podemos relacionar esta tabla de ventas con la tabla de artículos por el id_artículo y con la de locales por el id_local. Con esto tendríamos un solo DataFrame con toda la información que nos han proporcionado 

```python
#Primero añadimos a las ventas la información de cada tienda:
df_franquicia = df_ventas.merge(
                df_final,
                left_on = "id_local",
                right_index = True,
                how= "left" 
)
```

```python
#Y ahora le agramos la descripción de cada artículo:
df_franquicia = df_franquicia.merge(
                df_arti,
                left_on = "id_producto",
                right_index = True,
                how= "left"
)
```

```python
df_franquicia.info()
```

```python
print(f"Nuestro dataframe de FINAL DE VENTAS consta de {df_franquicia.shape[0]} filas y {df_franquicia.shape[1]} columnas")
print(f"El volumen de datos que contiene nuestro dataframe es: {df_franquicia.size}")
```

Tenemos un solo DataFrame de más de 1.2Millones de registros con 36 columnas, en total +4.5millones de datos, haremos ahora una visión general de los datos

```python
df_franquicia.describe(percentiles = None).T
```

Aquí ya podemos extraer algunas conclusiones:  
-El alquiler medio es de unos 5.5K pero hay bastante variación, ya que van desde los 800€ hasta casi 10K.  
-El primer local se abrió en Enero de 2019 y el último a finales de 2025.  
-Los Royalties tienen cierta variación, entre el 4% y el 7%, con un promedio de 5.5%  
-El tamaño estándar de los locales es de unos 260 metros cuadrados  
-Hay franquiciados que tienen un solo local y alguno cuenta hasta con 9 locales  
-Los precios de venta están en un rango de 1.75€ hasta 8.50€


```python
# Veamos las ventas por provincia
tabla_vtas_prov = (
    df_franquicia.groupby("Provincia") #Agrupamos los datos por provincia
    .agg(tot_ventas=("importe", "sum"))#sumamos las ventas
    .round(2) #readondeamos a 2 decimales
    .sort_values("tot_ventas", ascending=False) # y ordenamos de mayor a menor
)
tabla_vtas_prov["%_ventas"] = (
    tabla_vtas_prov["tot_ventas"] / tabla_vtas_prov["tot_ventas"].sum() * 100
).round(1)

tabla_vtas_prov
```

Madrid concentra más de la cuarta parte de las ventas, le siguen las provincias de Barcelona y Málaga.  
Veamos ahora como se distribuyen las ventas por familias de productos

```python
tabla_vtas_fami = (
    df_franquicia.groupby("familia") 
    .agg(tot_ventas=("importe", "sum"))
    .round(2) 
    .sort_values("tot_ventas", ascending=False)
)
tabla_vtas_fami["%_ventas"] = (
    tabla_vtas_fami["tot_ventas"] / tabla_vtas_prov["tot_ventas"].sum() * 100
).round(1)

tabla_vtas_fami
```

Claramente se venden muchas más bebidas que comida. Aquí hay una interesante oportunidad de negocio. Profundicemos en el detalle de las ventas por artículos

```python
tabla_vtas_cate = (
    df_franquicia.groupby("categoria") 
    .agg(tot_ventas=("importe", "sum"))
    .round(2) 
    .sort_values("tot_ventas", ascending=False)
)
tabla_vtas_cate["%_ventas"] = (
    tabla_vtas_cate["tot_ventas"] / tabla_vtas_prov["tot_ventas"].sum() * 100
).round(1)

tabla_vtas_cate
```

El principal negocio de esta empresa es la venta de CERVEZA. Tienen una gran oportunidad de crecimiento si potencian la venta de comida con promociones cruzadas de Bebida + Comida.  
Dada la importancia de esta categoría, vamos a ver cueles son las marcas y formatos que mejor funcionan

```python
df_cervezas = df_franquicia[df_franquicia["categoria"] == "Cerveza"]
```

```python
matriz_cervezas = df_cervezas.pivot_table(
    values="importe",
    index="marca",
    columns="formato",
    aggfunc="sum",
    fill_value=0
)
matriz_cervezas.style.format(lambda x: f"{x:,.0f}".replace(",", "X").replace(".", ",").replace("X", "."))
```

Vemos que la distribución por marcas esta bastante proporcionada, no hay ninguna que destaque claramente. Sin embargo, en cuanto a los formatos, vemos que dentro del formato BOTELLA, el tercio predomina sobre el botellín y la Jarra sobre el Doble en el los formatos de BARRIL. Probablemente esto atienda a la diferencia de PVP, para comprobarlo, vamos a generar la misma matriz pero con las CANTIDADES

```python
matriz_cervezas = df_cervezas.pivot_table(
    values="cantidad",
    index="marca",
    columns="formato",
    aggfunc="sum",
    fill_value=0
)
matriz_cervezas.style.format(lambda x: f"{x:,.0f}".replace(",", "X").replace(".", ",").replace("X", "."))
```

Efectivamente las cantidades son muy similares, por lo que la diferencia de facturación viene dada por un mayor precio de venta

Veamos cuales on los franquiciados que aglutinan más ventas

```python
tabla_vtas_fq = (
    df_franquicia.groupby("nombre") 
    .agg(tot_ventas=("importe", "sum"))
    .round(2) 
    .sort_values("tot_ventas", ascending=False)
)
tabla_vtas_fq["%_ventas"] = (
    tabla_vtas_fq["tot_ventas"] / tabla_vtas_prov["tot_ventas"].sum() * 100
).round(1)

tabla_vtas_fq
```

Vamos a ver como se reparten las ventas a lo largo del año

```python
df_franquicia["año"] = df_franquicia["fecha"].dt.year
df_franquicia["mes"] = df_franquicia["fecha"].dt.month

matriz_mensual = df_franquicia.pivot_table(
    values="importe",
    index="mes",
    columns="año",
    aggfunc="sum",
    fill_value=0
)
matriz_mensual
```
