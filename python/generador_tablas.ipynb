```python
import pandas as pd
import numpy as np

n_franquiciados = 20
franquiciados = pd.DataFrame({
    "id_franquiciado": range(1, n_franquiciados+1),
    "nombre": ["Franquiciado_" + str(i) for i in range(1, n_franquiciados+1)],
    "tipo_contrato": np.random.choice(["Gold", "Silver", "Basic"], n_franquiciados),
    "tipo_franquiciado": np.random.choice(["Inversor", "Autoempleo", "Propio"], n_franquiciados),
    "royalty": np.random.uniform(0.05, 0.12, n_franquiciados)
})
```

```python
franquiciados.to_csv("data/franquiciados.csv", index=False)
```

```python
n_locales = 76
locales = pd.DataFrame({
    "id_local": range(1, n_locales+1),
    "nombre_local": ["Cervecería_" + str(i) for i in range(1, n_locales+1)],
    "id_franquiciado": np.random.choice(franquiciados["id_franquiciado"], n_locales),
    #"codigo_postal": np.random.choice(lista_codigos_postales, n_locales)
})
```

```python
locales.to_csv("data/locales.csv", index=False)
```

```python
n_productos = 25
categorias = ["Cerveza", "Aperitivo", "Refresco"]
marcas = ["Mahou", "Estrella Galicia", "Alhambra", "Heineken"]
formatos = ["Caña", "Jarra", "Botellín", "Tercio"]

productos = pd.DataFrame({
    "id_producto": range(1, n_productos + 1),
    "categoria": np.random.choice(categorias, n_productos),
    "marca": np.random.choice(marcas, n_productos),
    "formato": np.random.choice(formatos, n_productos),
    "precio_venta": np.random.uniform(1.5, 6.0, n_productos)
})
```

```python
productos.to_csv("data/productos.csv",index=False)
```

```python
fechas = pd.date_range("2024-01-01", "2024-06-30", freq="D")

ventas_2024S1 = []

for fecha in fechas:
    for local in locales["id_local"]:
        for prod in productos["id_producto"]:
            unidades = np.random.poisson(lam=3)  # media de 3 unidades/día
            if unidades > 0:
                precio = productos.loc[productos.id_producto == prod, "precio_venta"].values[0]
                ventas_2024S1.append([fecha, local, prod, unidades, unidades * precio])

ventas_2024S1_df = pd.DataFrame(ventas_2024S1, columns=["fecha", "id_local", "id_producto", "unidades", "importe"])
```

```python
ventas_2024S1_df.to_csv("data/ventas_2024S1.csv",index=False)
```
